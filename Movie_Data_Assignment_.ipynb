### **Question 26**
(Visualization) Dataset mein 'Certificate' wise average 'Gross' kamai ka ek Bar Plot banayein.

```python
gross_num = pd.to_numeric(movie['Gross'].astype(str).str.replace(',', '', regex=False), errors='coerce')
avg_gross = movie.assign(Gross_num=gross_num).groupby('Certificate')['Gross_num'].mean().dropna().sort_values(ascending=False)
avg_gross.plot(kind='bar', edgecolor='black')
plt.title('Average Gross by Certificate')
plt.xlabel('Certificate')
plt.ylabel('Average Gross')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### **Question 30**
Dataset ko poora clean karne ke baad safely ek nayi csv file 'cleaned_movies.csv' ke naam se save karein.

```python
cleaned_movie = movie.copy()
cleaned_movie['Runtime'] = pd.to_numeric(cleaned_movie['Runtime'].astype(str).str.replace(' min', '', regex=False), errors='coerce')
cleaned_movie['Gross'] = pd.to_numeric(cleaned_movie['Gross'].astype(str).str.replace(',', '', regex=False), errors='coerce')
cleaned_movie['Meta_score'] = pd.to_numeric(cleaned_movie['Meta_score'], errors='coerce')
cleaned_movie['Released_Year'] = pd.to_numeric(cleaned_movie['Released_Year'], errors='coerce')
cleaned_movie['Gross'] = cleaned_movie['Gross'].fillna(cleaned_movie['Gross'].median())
cleaned_movie.to_csv('cleaned_movies.csv', index=False)
print('File saved as cleaned_movies.csv')
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
plot_data = movie[movie['Genre'].str.contains('Drama|Action', case=False, na=False)].copy()
plot_data['Main_Genre'] = plot_data['Genre'].apply(lambda x: 'Drama' if 'Drama' in x else ('Action' if 'Action' in x else None))
plot_data['Gross_num'] = pd.to_numeric(plot_data['Gross'].astype(str).str.replace(',', '', regex=False), errors='coerce')
plot_data = plot_data.dropna(subset=['Gross_num', 'Main_Genre'])
plot_data.boxplot(column='Gross_num', by='Main_Genre')
plt.title('Gross Revenue: Drama vs Action')
plt.suptitle('')
plt.xlabel('Genre')
plt.ylabel('Gross Revenue')
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = movie.nlargest(5, 'IMDB_Rating').copy()
meta = pd.to_numeric(top5['Meta_score'], errors='coerce')
plot_df = pd.DataFrame({'Movie': top5['Series_Title'], 'IMDB Rating (x10)': top5['IMDB_Rating'] * 10, 'Meta Score': meta})
plot_df = plot_df.set_index('Movie')
plot_df.plot(kind='bar', figsize=(10, 5), edgecolor='black')
plt.title('Top 5 Rated Movies: IMDB vs Meta Score')
plt.ylabel('Score out of 100')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
